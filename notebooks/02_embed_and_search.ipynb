```python
import numpy as np
import pandas as pd

df = pd.read_csv("../data/clean.csv")
emb = np.load("../data/embeddings.npy")
pairs = pd.read_csv("../data/remake_pairs.csv")

def row_of(title, year):
    return df.index[(df["Title"] == title) & (df["Release Year"] == year)][0]

rng = np.random.default_rng(0)
random_scores = []
for _ in range(2000):
    i, j = rng.integers(0, len(df), size=2)
    random_scores.append(emb[i] @ emb[j])
print("average similarity of random film pairs:", round(float(np.mean(random_scores)), 3))

for _, r in pairs.iterrows():
    a = row_of(r["original_title"], r["original_year"])
    b = row_of(r["remake_title"], r["remake_year"])
    print(r["original_title"], "->", r["remake_title"], round(float(emb[a] @ emb[b]), 3))
```
